```python
# can use to work with tabular data
import pandas as pd
df = pd.read_csv("competition.csv")
df.head() # show the first rows
df.describe() #show some basic statistics


```

```python
import matplotlib.pyplot as plt #matplotlib helps to creates charts and visualizations
df.plot(kind="bar", x="Name", y="Age", legend=False)
plt.title("Age per Athlete")
plt.ylabel("Age of practice")
plt.show()
```

```python
df.plot (kind="pie", y="Age", labels=df["Name"], legend=False, autopct="%1.1f%%")
plt.title("Age of athletes")
```

```python
plt.scatter(df["Age"], df ["Years_of_practice"], s=df["Years_of_practice"]*60)
#s=df["Years_of_practice"]*60 this means if student has ,ore years of practice their bubble is bigger
#alpha=0.6 makes the bubble bit transparent
plt.xlabel("Age")
plt.ylabel("Years of practice" )
plt.title("Age vs years of practice")
```

```python
import seaborn as sns #use this to create more clean and formatted plots
sns.boxplot(data=df, y="Age")
plt.title("distribution of athlete age")
plt.show()
```

```python
plt.figure(figsize=(8,5))
df[["Age", "Years_of_practice"]].boxplot()
plt. title("Box plot for age and years of practice")
plt.show
```

```python
avg_age= df.groupby("Sport")["Age"].mean()
avg_age.plot (kind="bar")
plt.title ("Average years of practice per athletes")
plt.show()
```
